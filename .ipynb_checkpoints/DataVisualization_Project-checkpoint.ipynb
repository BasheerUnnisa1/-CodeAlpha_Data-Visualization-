```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sales_data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'Sales': [1200, 1900, 3000, 2500, 3200, 4100]
}

df = pd.DataFrame(sales_data)

df
```

```python
plt.figure(figsize=(8,5))

sns.lineplot(x='Month', y='Sales', data=df, marker='o')

plt.title("Monthly Sales Analysis")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.barplot(x='Month', y='Sales', data=df)

plt.title("Monthly Sales Bar Chart")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

```python
print("Insights:")
print("- Sales increased overall from January to June.")
print("- Highest sales were recorded in June.")
print("- Lowest sales were recorded in January.")
print("- Strong growth can be seen after March.")
```

```python
plt.figure(figsize=(7,7))

plt.pie(
    df['Sales'],
    labels=df['Month'],
    autopct='%1.1f%%'
)

plt.title("Sales Distribution by Month")

plt.show()
```

```python
print("Conclusion:")
print("This project analyzed monthly sales data using Python.")
print("Different visualizations helped identify sales trends and performance.")
print("The analysis showed continuous business growth, especially after March.")
print("Data visualization makes decision-making easier and more effective.")
```
